```python
customer_info = customer_info.drop_duplicates()
print(customer_info.shape)
```

```python
print(customer_info['Geography'].value_counts())
```

```python
customer_info['Geography'] = customer_info['Geography'].replace({
    'French': 'France',
    'FRA': 'France'
})

print(customer_info['Geography'].value_counts())
```

```python
customer_info['EstimatedSalary']=customer_info['EstimatedSalary'].str.replace('€','').astype(float)
print(customer_info['EstimatedSalary'].value_counts())
```

```python
customer_info = customer_info[customer_info['EstimatedSalary']>0]
print(customer_info['EstimatedSalary'].min())
```
